```python
"""
Jayden Lewis
Maziar Sojoudian
COMP 2152 - 50322
101484621
03/02/2025
Assignment #1
"""



# Part B
gym_member = "Alex Alliton" # str
preferred_weight_kg = 20.5  # float
highest_reps = 25           # int
membership_active = True    # bool



# Part C
# Dictionary with tuples defining time on yoga, running, and weightlifting
workout_stats = {
    "Alex": (30, 45, 20),
    "Jamie": (25, 50, 30),
    "Taylor": (40, 35, 45)
}



# Part D
# Calculating the total workout time (in minutes) for each person
updated_workout_stats = workout_stats.copy()
for name, times in workout_stats.items():
    total = sum(times)
    updated_workout_stats[f"{name}_total"] = total
workout_stats = updated_workout_stats



# Part E
# Nested list of workout_stats
# This is a list of lists. Each sublist has the workout data for each person
workout_list = [list(values) 
                for key, values in workout_stats.items() 
                if isinstance(values, tuple)]



# Part F
# Displaying yoga and running data for each person from the nested list
# Only displaying weightlifting data for the last 2 people
names = list(workout_stats.keys())

print("=== Workout Times ===")

for i, row in enumerate(workout_list):
    
    if i < len(workout_list) - 2:
        print(f"{names[i]}: " +
              f"Yoga time = {row[0]}, " +
              f"Running time = {row[1]}")
    
    else:
        print(f"{names[i]}: " +
              f"Yoga time = {row[0]}, " +
              f"Running time = {row[1]}, " +
              f"Weightlifting time = {row[2]} min")



# Part G
for i, name in enumerate(names):

    total_time = updated_workout_stats.get(f"{name}_total", 0)
    
    if total_time >= 120:
        print(f"\nGreat job staying active, {name}!\n")



# Part H
found = False

print("=== Search for a user ===")
search = input("Please enter their name: ")
for i in workout_stats:
    if search.lower() == i.lower():
        
        print(f"User {search} found!")
        
        print(f"Yoga: {workout_stats[i][0]} min\n"
              f"Running: {workout_stats[i][1]} min\n"
              f"Weightlifting: {workout_stats[i][2]} min\n"
              f"Total exercise time: {sum(workout_stats[i])} min")
        
        found = True
    
if found == False:
    print(f"User {search} not found.")

# Part I
# Extract total workout times only
totals = {name.replace("_total", ""): minutes for name, minutes in workout_stats.items() if "_total" in name}

# Find the friend with highest and lowest total workout time
max_friend, max_minutes = max(totals.items(), key=lambda x: x[1])
min_friend, min_minutes = min(totals.items(), key=lambda x: x[1])

# Print results
print("\n=== Total Workout Times ===")
print(f"Highest total workout time: {max_friend} with {max_minutes} minutes")
print(f"Lowest total workout time: {min_friend} with {min_minutes} minutes")

    
```
